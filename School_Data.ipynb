```python
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/CunyLaguardiaDataAnalytics/datasets/master/2014-15_To_2016-17_School-_Level_NYC_Regents_Report_For_All_Variables.csv')
```

```python
df.head(5)
```

```python
df['Mean Score'].dtype # many columns have decimal numbers, I want to change typ to float to be able to run aggregate functions
df['Mean Score'] = pd.to_numeric(df['Mean Score'],errors='coerce') # I've chosen to drop the missing values in the dataset
df['Mean Score'].dtype # type is now float instead of '0'
```

```python
bt_df = df[(df['School Name'] == 'Brooklyn Technical High School')]
bt_df.head(5)
```

```python
bt_df.groupby('Regents Exam') ['Mean Score'].mean().sort_values(ascending= False)
# Average highest subject is US History and Govt at 94.5 and lowest is Algebra/Trig at 76.85
```

```python
bt_df.groupby('Year') ['Mean Score'].mean().sort_values(ascending= False)
# Average score is similar across 3 years, highest is 86.7 in 2017 and lowest is 84.56 in 2016
```

```python
st_df = df[(df['School Name'] == 'Stuyvesant High School')]
st_df.head()
```

```python
st_df.groupby('Regents Exam') ['Mean Score'].mean().sort_values(ascending= False)
# Average highest subject is US History and Govt at 96.4 and lowest is Algebra 2 at 85.73
```

```python
st_df.groupby('Year') ['Mean Score'].mean().sort_values(ascending= False)
# Average score is almost identical from year to year, highest score was in 2015 at 91.98, lowest was 2016 at 91.21
```

```python
df2= df[df['School Name'].isin(['Brooklyn Technical High School','Stuyvesant High School'])]
df2.head(5)
```

```python
df2[df2['Regents Exam'] == 'Common Core Algebra'].groupby(['School Name', 'Year'])['Mean Score'].mean().sort_values(ascending= False).plot(
    kind='bar', xlabel= 'School Name and Year', ylabel= 'Score', title='Average Common Core Algebra Score by School and Year', color= 'purple')
```

I chose to do an analyisis on Stuyvenant and Brooklyn Tech because Stuyvesant is the most prestigious high school in NYC and Brooklyn Tech is the largest.

As I expected, Stuyvesants average test scores were higher at around 91, but Brooklyn Tech's scores are also great at around 85 each year.

Both schools had a similar breakdown of which subjects had the highest average score. The humanities had higher scores and most of the stem exams had lower scores.
