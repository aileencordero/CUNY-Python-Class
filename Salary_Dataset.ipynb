```python
import pandas as pd
df=pd.read_csv('https://raw.githubusercontent.com/CunyLaguardiaDataAnalytics/datasets/master/Salaries.csv')
```

```python
#Any null values? No
df.isnull().sum().sum()
```

```python
df.head()
```

```python

#Data Aggregation
#Mean is $113,706.46
salaries=df['salary']
mean=salaries.mean()
mean

#median is $107,300
salaries=df['salary']
salaries.median()


#median by gender is female $103,750 male is $108,043
salaries=df['salary']
salaries.groupby(df['sex']).median()


#median by job position Professor is $123321.50, Associate professor is $95626.50, Assistant professor is $79,800
salaries=df['salary']
salaries.groupby(df['rank']).median()
```

```python
df.groupby('yrs.service')['salary'].median().sort_values(ascending=False).head(5).plot(
    kind='bar', ylabel='Salary', xlabel='Years Teaching', title='Top 5 Median Salaries by Years Worked')
```

```python
df.groupby('sex')['salary'].median().plot(kind='bar', ylabel='Salary', xlabel='Sex', title='Median Income by Gender', color='pink')
```

```python
df2=df[['rank', 'yrs.since.phd', 'salary']]
```

```python
df2.groupby('yrs.since.phd')['salary'].median().sort_values(ascending=False).head(5).plot(
    kind='bar', ylabel='Salary', xlabel='Years Since PhD', title='Top 5 Median Salaries by Years Since PhD', color='purple')
```
