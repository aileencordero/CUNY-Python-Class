```python
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/CunyLaguardiaDataAnalytics/datasets/master/mtcars.csv')
```

```python
df = df.rename(columns= {'Unnamed: 0': 'Car Model'})
```

```python
df2 = df[['mpg', 'cyl', 'hp', 'wt']]
```

```python
df2 = df2.rename(columns= {'mpg' : 'miles per gallon', 'cyl' : 'cylinder', 'hp' : 'horse power', 'wt' : 'weight'})
```

```python
df2.head()
```

```python
# summary for mpg
average_miles = df2['miles per gallon'].mean()
average_miles # average mpg is 20.09

median_miles = df2['miles per gallon'].median()
median_miles # median mpg is 19.2
```

```python
avg_cyl = df2['cylinder'].mean()
avg_cyl # average cylinder is 6.1875

med_cyl = df2['cylinder'].median()
med_cyl # median cylinder is 6
```

```python
avg_hp = df2['horse power'].mean()
avg_hp # average horse power is 146.6875

med_hp = df2['horse power'].median()
med_hp #median horse power is 123
```

```python
avg_weight = df2['weight'].mean()
avg_weight #average is 3.217

med_weight = df2['weight'].median()
med_weight #median is 3.325
```
